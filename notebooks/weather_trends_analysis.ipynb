# Weather Trends Analysis\n\nAnalyze historical weather data to identify trends, seasonality, and basic statistics.

```python
import pandas as pd\nimport numpy as np\nimport matplotlib.pyplot as plt\nimport seaborn as sns\n%matplotlib inline\nsns.set_style('whitegrid')
```

```python
df = pd.read_csv('../data/weather_daily.csv')\ndf['date'] = pd.to_datetime(df['date'])\ndf.head()
```

```python
df.describe()
```

```python
plt.figure(figsize=(14,5))\nsns.lineplot(x='date', y='avg_temp_c', data=df)\nplt.title('Daily Average Temperature')\nplt.show()
```

```python
plt.figure(figsize=(10,6))\nsns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')\nplt.title('Correlation Matrix')\nplt.show()
```

## Insights\n- Temperature follows seasonal pattern over time.\n- Humidity and precipitation show expected relationships.\n- Correlations align with typical weather dynamics.
